Visualization 1: I decided to look at the average amount of tournament wins each seed has historically had. Although it was likely obvious that the general trend would just be 1-16 (1 being most wins, 16 being lowest), I wanted to see what the difference was between the "same tier" of seed (1 and 2, 3 and 4 etc.) to see how big of a difference there was between being on the direct opposite side of the bracket. 

```r
avg_tourney_wins_by_seed <- tidy_march_madness |>
    filter(seed > 0) |>
    group_by(seed) |>
    summarize(avg_tourney_wins = mean(tourney_wins)) #I wanted to find the average amount of wins that each seed gets each tournament.
seed_vs_tourney_wins <- ggplot(avg_tourney_wins_by_seed, aes(x = factor(seed), y = avg_tourney_wins)) +
       geom_bar(stat = "identity", fill = "blue", colour = "white") +
       labs(title = "Average Tournament Wins by Seed", x = "Seed", y = "Average Tournament Wins")
seed_vs_tourney_wins
```

Visualization 2: A known concept in sports is referred to as "Home Court Advantage", which essentially means a team is more likely to win at home than on the road. This being said, I wanted to see how big of an advantage it was to have the first game of the tournament at home vs on the road, and theoretically, it should be rather significant if the concept is actually real.

```r
home_game_adv <- tidy_march_madness |>
    group_by(first_game_at_home) |>
    summarize(mean_tourney_wins = mean(tourney_wins))

home_start_vs_tourney_wins <- home_game_adv |>
    ggplot(aes(x = first_game_at_home, y = mean_tourney_wins)) +
    geom_bar(stat = "identity", fill = "blue", alpha = 0.8) +
    labs(title = "Average Tournament Wins by First Game Home Court Advantage", x = "Home Court Advantage (Y = yes or N = no)", y = "Average Tournament Wins")
home_start_vs_tourney_wins
```
